# Types of Inheritance

## Single Inheritance

- When a Child class(derived class) inherits from exactly one parent class (base class).
- Simplest form of inheritence.
- Consturctor of parent class will be called first and then child class.
- and opposite for deconstructor.






```python
#include <iostream>
using namespace std;

class Human{
    protected:
    string name;
    int age;

    public:
    Human(string name, int age){
        this->name=name;
        this->age=age;
    }
    void work(){
        cout<<"Currently i am working\n"<<endl;
    }
};
class Student: public Human{
    int roll;
    int fees;
    public:
    Student(string n, int a, int r, int f):Human(n, a), roll(r), fees(f){}
    void Display(){
        cout<<"Name: "<<name<<" Age: "<<age<<" Roll no.: "<<roll<<" Fees: "<<fees<<endl;
    }
};
int main(int argc, char const *argv[])
{
    Student A1("Aryan", 18, 49, 2000);
    A1.work();
    A1.Display();
    return 0;
}
// output:
//  Currently i am working

// Name : Aryan Age : 18 Roll no. : 49 Fees : 2000
```

### Question:
- Create a base class Animal with a public method eat() that prints “Animal is eating.” Derive a class Dog from Animal. The Dog class must override the eat() method to print “Dog is eating kibble.” Additionally, add a new public method bark() to the Dog class that prints “Woof! Woof!”.

```python
#include <iostream>
using namespace std;

class Animal{
    public:
    void eat(){
        cout<<"Animal is eating generic food"<<endl;
    }
    void sleep(){
        cout<<"Animal is sleeping"<<endl;
    }
};
class Dog:public Animal{
    public:
    void eat(){
        cout<<"Dog is eating kibble"<<endl;
    }
    void bark(){
        cout<<"Woof! Woof!"<<endl;
    }
};
int main(){
    Dog a;
    a.eat();
    a.bark();
    a.sleep();
}
```

## Multilevel Inheritance:

- Multilevel inheritance occurs when a class is derived from another derived class, creating a chain of inheritance. It represents an "is-a" relationship across multiple layers (e.g., Grandparent → Parent → Child).




{height=50px}

```python
#include <iostream>
using namespace std;

class Person{
    protected:
    string name;

    public:
    void introduce(){
        cout<<"hello myself"<<name<<endl;
    }
};
class employee: public Person{
    protected:
    int salary;
    public:
    void monthly_salary(){
        cout<<"My monthly salary is "<<salary<<endl;
    }
};
class manager:public employee{
    public:
    string department;
    public:
    manager(string name, int salary, string department){
        this->department=department;
        this->name=name;
        this->salary=salary;
    }
    void work(){
        cout<<"I am leading "<<department<<endl;
    }
    void display(){
        cout<<"Name: "<<name<<endl;
        cout<<"Department: "<<department<<endl;
        cout<<"Salary: "<<salary<<endl;
    }
};
int main(int argc, char const *argv[])
{
    manager a("Aryan", 30000, "IT");
    a.display();
    a.work();
    a.monthly_salary();
    a.introduce();
    return 0;
}


```

## Multiple Inheritance:

- occurs when a child class (derived class) inherits from two or more independent parent classes (base classes).
- This allows child class to combine behaviours and attributes from various sources.
- **`the constructors are executed in the exact order they are listed in the class declaration line (from left to right), regardless of how you arrange them in your constructor initializer list.`**




```python
#include <iostream>
using namespace std;

class Engineer{
    public: 
    string specialisation;

    void work(){
        cout<<"I have specialisation in "<<specialisation<<endl;
    }
};
class Single{
    public:
    string time;
    void relationship(){
        cout<<"I am single for "<<time<<endl;
    }
};
class Weirdo{
    public:
    string weird;

    void is_weirdo(){
        cout<<"Am i weird? "<<weird<<endl;
    }
};
class me: public Engineer, public Single, public Weirdo{
    protected:
    string name;
    public:
    me(string specialisation, string time, string weird, string name){
        this->name=name;
        this->specialisation=specialisation;
        this->weird=weird;
        this->time=time;
    }
    void myself(){
        cout<<"yes, me "<<name<<", I am weird af"<<endl;
    }
};
int main(int argc, char const *argv[])
{
    me a("AIML","Lifetime", "Yes", "Aryan");
    a.is_weirdo();
    a.myself();
    a.relationship();
    a.work();
    return 0;
}


```

## Hierarchial inheritance:

- "Hierarchical Inheritance is a type of inheritance where multiple child classes inherit from a single parent class."





## Hybrid Inheritance:

- Hybrid Inheritance is a combination of two or more types of inheritance (such as Single, Multiple, Hierarchical, or Multilevel inheritance) within a single program.

- It is used to model complex, real-world relationships where a single inheritance pattern is not enough.



## Multipath Inheritance: 

- Multipath Inheritance is a special, specific case of hybrid inheritance where a child class inherits from multiple parent classes, which themselves share a common base class.
