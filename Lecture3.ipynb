## Static Data Member
  - They are attribute of classes or class member.
  - It is Declared using `static` keyword.
  - Only one copy of that member is created for the enitre class and is shared by all the object.
  - It is initialised before any object of this class is created.

```python
#include <iostream>
using namespace std;
class Customer
{
    public:
    string name;
    int acc_no, balance;
    static int total;
    static int total_cash;

public:
    Customer(string name, int acc_no, int balance)
    {
        this->name = name;
        this->acc_no = acc_no;
        this->balance = balance;
        total++;
        total_cash += balance;
    }
    void Display(){
        cout << "Name : " << name << " Acc No. : " << acc_no << " Balance: " << balance <<endl;
    }
    void static displaystatic(){
        cout << "Total Customer is : " << total << endl;
        cout << "Total balance is : " << total_cash << endl;
    }
    void deposit(int balance){
        this->balance += balance;
        total_cash += balance;
    }
};
int Customer::total = 0;
int Customer::total_cash = 0;
int main()
{
    Customer a("Aryan", 107, 10000);
    Customer b("Tanmay", 108, 20000);
    Customer c("Agrim", 109, 30000);
    Customer d("shivesh", 110, 50000);
    a.Display();
    b.Display();
    c.Display();
    d.Display();
    Customer::displaystatic();
    a.deposit(200);
    Customer::displaystatic();
}
```

## const Keyword (homework)
- A member function declared with const at the end of its prototype cannot modify any non-static data members of the class. These are frequently used for "getter" methods.
   - Syntax: void display() const; 
   - Rule: If you attempt to change a variable inside this function, the compiler throws an error.

```python
class Account {
private:
    double balance;
public:
    Account(double b) : balance(b) {}

    // Const member function: Read-only access
    double getBalance() const {
        // balance = 0; // Error: Cannot modify member variables here
        return balance;
    }
};
```

### Example: Circle Class With Constant Methods
- Implement a Circle class with a private member double radius. Add a constructor to initialize the radius. Include a public constant member function get_area() that calculates and returns the area of the circle (pir, use pi = 3.14159)

```python
#include<iostream>
#include<math.h>
using namespace std;
#define pi 3.14159
class Circle{
    private:
        double radius;
    public:
    Circle(double r):radius(r){}
    double get_area(){
        return pi * pow(radius, 2);
    }
};

int main(int argc, char const *argv[])
{
    double r;
    cout << "Enter radius: ";
    cin >> r;
    Circle C(r);
    cout << "Circle with radius " << r << " has an area of " << C.get_area();
}
```

## Encapsulation
   - Wrapping up data and information in a single unit, while controlling access to them. 

```python
#include <iostream>
#include <string>

class BankAccount {
private:
    // Hidden data members (Data Hiding)
    std::string accountHolder;
    double balance;

public:
    // Constructor to initialize data securely
    BankAccount(std::string name, double initialBalance) {
        accountHolder = name;
        if (initialBalance >= 0) {
            balance = initialBalance;
        } else {
            balance = 0; // Prevent invalid initial state
        }
    }

    // Public method to modify private data with strict business logic
    void deposit(double amount) {
        if (amount > 0) {
            balance += amount;
            std::cout << "Deposited $" << amount << " successfully.\n";
        } else {
            std::cout << "Error: Deposit amount must be positive!\n";
        }
    }

    // Public method to modify private data with verification checks
    void withdraw(double amount) {
        if (amount <= 0) {
            std::cout << "Error: Withdrawal amount must be positive!\n";
        } else if (amount > balance) {
            std::cout << "Error: Insufficient funds!\n";
        } else {
            balance -= amount;
            std::cout << "Withdrew $" << amount << " successfully.\n";
        }
    }

    // Public getter method for safe read-only access
    double getBalance() const {
        return balance;
    }
};

int main() {
    // Create an encapsulated object
    BankAccount myAccount("Alice", 500.0);

    // Controlled modifications through public methods
    myAccount.deposit(150.0);
    myAccount.withdraw(100.0);
    myAccount.withdraw(600.0); // Triggers "Insufficient funds" protection

    // Reading the data safely
    std::cout << "Final Balance: $" << myAccount.getBalance() << "\n";

    // Direct access is strictly blocked by the compiler
    // myAccount.balance = 999999; // Error: 'balance' is a private member
    
    return 0;
}
```

### Example: Date Class with Validation
- Create a Date class with private members int day, int month, and int year. Implement a constructor that performs basic validation: if the month is not between 1 and 12, it should set the month to a default value (e.g., 1) and print an error message.
- This exercise highlights that constructors are not just for initialization but are also the ideal place for object validation.<br>By placing validation logic inside the constructor, we ensure that an object is never created in an invalid or inconsistent state. This principle contributes significantly to robust Encapsulation and object integrity.

```python
#include <iostream>

class Date {
private:
    int day;
    int month;
    int year;

public:
    // Parameterized Constructor
    Date(int d, int m, int y) 
        : day(d), month(m), year(y) 
    {
        // Validation Logic
        std::cout<< year << "-" << month << "-" << day << std::endl;
        if (month < 1 || month > 12) {
            std::cerr << "Warning: Invalid month (" << month << ") provided. Setting to 1." << std::endl;
            this->month = 1; // 'this->' is optional here but highlights the member being modified
        }
        // Basic validation on day could also be added (e.g., if day < 1 || day > 31)
    }

    void display() const {
        std::cout<<"Valid Date: "<<year << "-" << month << "-" << day << std::endl;
    }
};

int main() {
    std::cout<<"Date 1: ";
    Date valid_date(28, 10, 2025);
    valid_date.display();

    std::cout<<"Date 2: ";
    Date invalid_month_date(1, 13, 2025);
    invalid_month_date.display(); // Should show 2025-1-1

    return 0;
}
```

### Example: Book Class For Data Retrieval
- Design a Book class with private members std::string title, std::string author, and std::string isbn. Include a constructor to initialize them. Implement a public method get_details() that returns a single formatted std::string containing all the book’s information

```python
#include<iostream>
using namespace std;
class Book{
    private:
        string title;
        string author;
        string isbn;
    public:
    Book(string t, string a, string i):title(t), author(a), isbn(i){}
    string get_details(){
        return "Title: " + title + " | Author: " + author + " | ISBN: " + isbn;
    }

};

int main(int argc, char const *argv[])
{
    Book one("The C++ Guide", "Aryan", "109274902");
    cout << "Book record: " << endl;
    cout << one.get_details();
}
```
