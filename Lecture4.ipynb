## Inheritance

- The capability of a class to derive properties of another class: Parent and child.

```c
#include<iostream>
using namespace std;

class Human{
    private:
        int a;
    protected:
        int b;
    public:
        int c;
    void access(){
        a = 10;
        c = 10;
        b = 10;
    }
    // a,b,c are member variables of class Human which can be accessed by member functions of class Human. Does'nt matter whether they are private, protected or public. But outside the class only public member can be accessed. Private and protected members can only be accessed by member functions of the class or derived classes.
};
int main(int argc, char const *argv[])
{
    Human Aryan;
    Aryan.c = 10; //only public member can be accessed outside the class
    return 0;
}
```

```c
#include<iostream>
using namespace std;

class Human{
    public:
    string name;
    int age, weight;
};
class Student: public Human{ 
    public:
    int roll_number, fees;
};
int main(int argc, char const *argv[])
{
    Student A;
    A.name = "Aryan";
    A.age = 10;
    A.weight = 80;
    A.roll_number = 01;
    A.fees = 100;
}
```
