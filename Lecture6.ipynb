# Types of Inheritance

## Single Inheritance

- When a Child class(derived class) inherits from exactly one parent class (base class).
- Simplest form of inheritence.
- Consturctor of parent class will be called first and then child class.
- and opposite for deconstructor.






```python
#include <iostream>
using namespace std;

class Human{
    protected:
    string name;
    int age;

    public:
    Human(string name, int age){
        this->name=name;
        this->age=age;
    }
    void work(){
        cout<<"Currently i am working\n"<<endl;
    }
};
class Student: public Human{
    int roll;
    int fees;
    public:
    Student(string n, int a, int r, int f):Human(n, a), roll(r), fees(f){}
    void Display(){
        cout<<"Name: "<<name<<" Age: "<<age<<" Roll no.: "<<roll<<" Fees: "<<fees<<endl;
    }
};
int main(int argc, char const *argv[])
{
    Student A1("Aryan", 18, 49, 2000);
    A1.work();
    A1.Display();
    return 0;
}
// output:
//  Currently i am working

// Name : Aryan Age : 18 Roll no. : 49 Fees : 2000
```

### Question:
- Create a base class Animal with a public method eat() that prints “Animal is eating.” Derive a class Dog from Animal. The Dog class must override the eat() method to print “Dog is eating kibble.” Additionally, add a new public method bark() to the Dog class that prints “Woof! Woof!”.

```python
#include <iostream>
using namespace std;

class Animal{
    public:
    void eat(){
        cout<<"Animal is eating generic food"<<endl;
    }
    void sleep(){
        cout<<"Animal is sleeping"<<endl;
    }
};
class Dog:public Animal{
    public:
    void eat(){
        cout<<"Dog is eating kibble"<<endl;
    }
    void bark(){
        cout<<"Woof! Woof!"<<endl;
    }
};
int main(){
    Dog a;
    a.eat();
    a.bark();
    a.sleep();
}
```
