# What is OOPS?
 - It is an approach or programming pattern where the programs are structured around objects rather than functions and logic.

```c
string name;
int age, roll_number;
string grade;
cin>>name>>age>>roll_number>>grade;
// suppose we need to use it multiple times,..
// we make a fuctions to call it multiple times
void f(){
    string name;
    int age, roll_number;
    string grade;
    cin>>name>>age>>roll_number>>grade;
}
```

## Access modifier
   - Access modifier or access specifier are those keywords which are used for deciding the accessibility and visibility of data(variables) and function code in a class to other part of the codes.
   - ### Public:
        - can be access inside or from outside of the class.
   - ### Private:
        - This is the most secure one.
        - This can be access only to those function or data which are inside the same class.
        - This can't be changed or touched from outside the class.
   - ### Protected:
        - This is special one which is used during inheritance.
        - Only those class whose parent class is this, or only child/derived class can access to this protected code.     

```c
class BankAccount {
public:
    string accountHolderName; // Koi bhi dekh sakta hai

private:
    double balance;           // Security zaroori hai! Bahar ka code ise direct change nahi kar sakta
};
```

### So What is Class?

    - It is a user defined data type.
    - Blueprint for creating object.
    - Data variables inside this is calsed `Attributes` and Function is known as `Methods`.

```c
//setter and getter functions
class Student{
    private:
    string name;
    int age, roll_number;
    string grade;
    public:
    void setname(string n){
        name = n;
    }
    void setage(int a){
        age = a;
    }
    void setroll_number(int r){
        roll_number = r;
    }
    void setgrade(string g){
        grade = g;
    }
    int getage(){
        return age;
    }
    int getroll_number(){
        return roll_number;
    }
    int getgrade(){
        return grade;
    }
    string getname(){
        return name;
    }
}
Student s1;
s1.setname("John");
s1.setage(20);
s1.setroll_number(101);
s1.setgrade("A");

```

## What is Object??
  - It is an entity that has a state and Behaviour.
  - Anything that exits in the physical world.

  - ### Size of an object:
       - It depends on the class. Object size depends on how much memory the class will take.

```c
class Student{
    public:
    char a;
    int age, roll_number;
    char c;
};
// consider 32 bit system
// size would be 1 + 4 + 4 + 1 = 10 bytes
// but due to padding it would be 12 bytes i.e., 12 bytes= 3*4 bytes
// padding is done to make the size of the class a multiple of the largest data type in the class
// in this case the largest data type is int which is 4 bytes
```

   - ### Structure / Class Padding or Data Alignment:
       - Structure or Class Padding is a mechanism where the compiler automatically inserts extra, empty bytes (called slack bytes) between structure members or at the end of the structure.
       - Processors (CPUs) do not read memory $1\text{ byte}$ at a time. Instead, they fetch data in fixed-sized chunks called a Memory Word. <br>
       $32\text{-bit}$ Architecture: $1\text{ Word} = 4\text{ bytes}$ <br>$64\text{-bit}$ Architecture: $1\text{ Word} = 8\text{ bytes}$
       - If a multi-byte data type (like a $4\text{-byte}$ integer) is misaligned (i.e., split across two word boundaries), the CPU has to perform two memory cycles and bit-shifting operations to read a single variable. This degrades performance.
       - The Alignment Rule: Every data type has an alignment requirement. A variable of size n bytes must be stored at a memory address that is a multiple of n.
           - char ($1\text{ byte}$): Can be stored at any address (multiples of 1).
           - short ($2\text{ bytes}$): Must be stored at addresses divisible by 2.
           - int / float ($4\text{ bytes}$): Must be stored at addresses divisible by 4.
           - double / long long ($8\text{ bytes}$): Must be stored at addresses divisible by 8.
       - #### How Compiler decides size?
           - Member Alignment: Each member must be aligned according to its size. Padding is added before a member if the current memory address layout is not a multiple of that member's size.
           - Total Structure Alignment: The total size of the structure/class must be a multiple of its largest scalar member's size (e.g., if the largest member is an int, total size must be a multiple of 4. If it's a double, a multiple of 8). This ensures that if you create an array of this structure, every element in the array remains perfectly aligned.
       - #### Reordering and memory consumptions:
            - ##### Case 1: Worst Case ordering: Alternating Sizes:
                - When small data types alternate with larger data types, maximum padding is introduced.
                - $$\text{Total Memory Consumed} = 1\text{ (a)} + 3\text{ (pad)} + 4\text{ (b)} + 1\text{ (c)} + 7\text{ (pad)} + 8\text{ (d)} = 24\text{ bytes}$$

```c
struct Case1 {
    char a;     // 1 byte
    int b;      // 4 bytes
    char c;     // 1 byte
    double d;   // 8 bytes
};
```

 - ##### Case 2 : Optimised Ordering( Descending Order of Size)
    -  By sorting the fields from largest to smallest, the smaller variables naturally fill up the remaining spaces inside the word boundary blocks.
    - $$\text{Total Memory Consumed} = 8\text{ (d)} + 4\text{ (b)} + 1\text{ (a)} + 1\text{ (c)} + 2\text{ (tail pad)} = 16\text{ bytes}$$

```c
struct Case2 {
    double d;   // 8 bytes
    int b;      // 4 bytes
    char a;     // 1 byte
    char c;     // 1 byte
};
```

- ##### Case 3: Nested Structures padding:
   - When a structure contains another structure as a member, the internal structure aligns based on its own largest element, not its total size.
   -  $$\text{Total Memory Consumed} = 1\text{ (a)} + 3\text{ (pad)} + 8\text{ (inner)} = 12\text{ bytes}$$

```c
struct Inner {
    char x;     // 1 byte + 3 bytes padding
    int y;      // 4 bytes
}; // sizeof(Inner) = 8 bytes. Largest element = 4 bytes.

struct Case3 {
    char a;       // 1 byte
    Inner inner;  // 8 bytes
};
```

- #### HOw to prevent padding? #pragma pack
   - If you are writing data directly to a file, sending raw packets over a network, or targeting low-memory embedded environments, you can explicitly disable compiler alignment rules using #pragma pack(1)
   

```c
#include <iostream>

#pragma pack(push, 1) // Force compiler to use 1-byte alignment boundary (No padding)
struct PackedStruct {
    char a;    // 1 byte
    int b;     // 4 bytes
    char c;    // 1 byte
};
#pragma pack(pop) // Restore original compiler settings

int main() {
    // Without packing, this would be 12 bytes.
    // With packing, it is exactly 1 + 4 + 1 = 6 bytes.
    std::cout << "Size: " << sizeof(PackedStruct) << " bytes"; 
    return 0;
}
```

#### Static vs Dynamic Memory Allocation


```c
class Student{
    public:
    string name;
    int age, roll_number;
    string grade;
}
Student *s= new Student;
s->name = "John";
s->age = 20;
s->roll_number = 101;
s->grade = "A";
// or
*s.name = "John";
*s.age = 20;    
*s.roll_number = 101;
*s.grade = "A";
```
