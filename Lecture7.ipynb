# Templates in C++

## What is a Template?

- A template is a feature in C++ that allows us to write generic programs that work with different data types without rewriting the same code for each type.
- Without templates, you might write separate functions

````python
int maxValue(int a, int b) {
    return (a > b) ? a : b;
}

float maxValue(float a, float b) {
    return (a > b) ? a : b;
}

double maxValue(double a, double b) {
    return (a > b) ? a : b;
}
// we can write instead:

template <typename T>
T maxValue(T a, T b) {
    return (a > b) ? a : b;
}
````

## Why do we use Template?

- <b>Code reusability:</b> Write a function or class once and use it with multiple data types.
- <b>Generic programming:</b> Develop code that works independently of a specific data type.
- <b>Reduced code duplication:</b> Avoid writing almost identical functions or classes.
- <b>Compile-time type safety:</b> The compiler checks whether the operations used in the template are valid for the supplied types.
- <b>Maintainability:</b> Changes to common logic need to be made in one place.

## Types of Template:
1. <b>Function Template</b><br>
Creates generic functions that operate on different data types.
Example: maxValue(), swapValues().
2. <b> Class Template</b><br>
Creates generic classes whose data members and member functions can work with different data types.
Example: Stack<int>, Stack<float>.

## Function Template:

- A function template is a blueprint for generating functions that perform the same operation on different data types.

````python
template <typename T>
return_type functionName(T parameter1, T parameter2) {
    // Function body
}
````

- We can also type class instead of typename in the typename parameter declaration.

````python
template <class T>
````

Both are equivalent in this context.

### Meaning of each part

| Component | Meaning |
|---|---|
| `template` | Declares a template. |
| `typename` | Introduces a type parameter. |
| `T` | A placeholder representing a data type. |
| `return_type` | The type returned by the function. |
| `functionName` | Name of the function. |

`T` is a conventional name, not a reserved keyword. You could use `Type`, `U`, or another valid identifier.

## 3.3 Example: Function Template to Find Maximum

**Complete program**

```cpp
#include <iostream>
using namespace std;

template <typename T>
T maxValue(T a, T b) {
    if (a > b)
        return a;
    else
        return b;
}

int main() {
    cout << maxValue(10, 20) << endl;
    cout << maxValue(4.5, 2.3) << endl;
    cout << maxValue('A', 'Z') << endl;

    return 0;
}
```

**Output**

```text
20
4.5
Z
```

### Dry run

| Function call | Type deduced for `T` | Result |
|---|---|---|
| `maxValue(10, 20)` | `int` | `20` |
| `maxValue(4.5, 2.3)` | `double` | `4.5` |
| `maxValue('A', 'Z')` | `char` | `Z` |

For the character example, the comparison uses the characters' underlying character-code values.

The compiler generates the appropriate function specialization for each required type.

## 3.4 Example: Function Template to Swap Values

```cpp
#include <iostream>
using namespace std;

template <typename T>
void swapValues(T &a, T &b) {
    T temp = a;
    a = b;
    b = temp;
}

int main() {
    int x = 10, y = 20;
    swapValues(x, y);
    cout << x << " " << y << endl;

    double p = 1.5, q = 2.5;
    swapValues(p, q);
    cout << p << " " << q << endl;

    return 0;
}
```

**Output**

```text
20 10
2.5 1.5
```

### Why use `T &a` and `T &b`?

The ampersand (`&`) declares reference parameters.

- They refer to the original variables.
- Changes inside the function affect the original variables.
- Without references, the function would swap only local copies.

**Exam point:** Templates support generic programming, while references allow this particular function to modify its arguments.

## 3.5 Explicit Template Arguments

Usually, the compiler determines `T` from the function arguments. This is called *template argument deduction*.

You can also specify the type explicitly:

```cpp
cout << maxValue<int>(10, 20);
cout << maxValue<double>(4, 8);
```

In the second call, the integer arguments are converted to `double` for the function parameters.

Explicit type arguments are useful when the type cannot be deduced or when you want to control it.

## 3.6 Function Template with Multiple Type Parameters

A template does not have to use only one type.

```cpp
#include <iostream>
using namespace std;

template <typename T, typename U>
void display(T a, U b) {
    cout << a << " " << b << endl;
}

int main() {
    display(10, 3.14);
    display('A', 100);

    return 0;
}
```

**Output**

```text
10 3.14
A 100
```

Here:

- `T` represents the first argument's type.
- `U` represents the second argument's type.

The types do not have to be identical.

# 4. Class Templates

## 4.1 Definition

A **class template** allows you to create a class that works with different data types.

For example, suppose you need a class that stores a single value. Without templates, you might create separate classes for integers, floating-point numbers, and characters. A class template lets you reuse the same class design for all these types.

## 4.2 Syntax

```cpp
template <typename T>
class ClassName {
private:
    T data;

public:
    void setData(T value) {
        data = value;
    }

    T getData() {
        return data;
    }
};
```

The class contains a placeholder `T`. When an object is created with a specific type, that type replaces `T` for the generated class specialization.

## 4.3 Example: Generic Box Class

**Complete program**

```cpp
#include <iostream>
using namespace std;

template <typename T>
class Box {
private:
    T value;

public:
    Box(T v) {
        value = v;
    }

    T getValue() {
        return value;
    }
};

int main() {
    Box<int> b1(100);
    Box<double> b2(12.5);
    Box<char> b3('A');

    cout << b1.getValue() << endl;
    cout << b2.getValue() << endl;
    cout << b3.getValue() << endl;

    return 0;
}
```

**Output**

```text
100
12.5
A
```

### Understanding the object declarations

| Declaration | Meaning |
|---|---|
| `Box<int> b1(100);` | Creates an integer Box object. |
| `Box<double> b2(12.5);` | Creates a double Box object. |
| `Box<char> b3('A');` | Creates a character Box object. |

Notice that `Box` is not an ordinary class name in these declarations. `Box<int>`, `Box<double>`, and `Box<char>` are different specializations of the same class template.

Each object stores its own `value`, and the chosen type determines the type of that member.

## 4.4 Class Template with Multiple Parameters

You can use multiple template parameters in a class.

```cpp
#include <iostream>
#include <string>
using namespace std;

template <typename T, typename U>
class Pair {
private:
    T first;
    U second;

public:
    Pair(T a, U b) {
        first = a;
        second = b;
    }

    void display() {
        cout << first << " " << second << endl;
    }
};

int main() {
    Pair<int, double> p1(10, 3.14);
    Pair<string, int> p2("Age", 18);

    p1.display();
    p2.display();

    return 0;
}
```

**Output**

```text
10 3.14
Age 18
```

The two template parameters can represent completely different types.

- `Pair<int, double>` stores an integer and a double.
- `Pair<string, int>` stores a string and an integer.

The `Pair` example also illustrates an OOP principle: data is private, and public member functions provide access to it. Templates and encapsulation work together.

# 5. Function Template vs Class Template

This is an important exam question.

| Function template | Class template |
|---|---|
| Used to create generic functions. | Used to create generic classes. |
| Performs an operation on arguments. | Defines data members and member functions. |
| Template types can often be deduced from function arguments. | Class template arguments generally need to be specified or deduced through supported class template argument deduction. |
| Example: `maxValue(2, 5)` | Example: `Box<int> b(10)` |
| Useful for reusable operations. | Useful for reusable data structures and objects. |

**Remember:** A function template is a blueprint for functions; a class template is a blueprint for classes.

# 6. Member Functions Defined Outside a Class Template

In exams, you may be asked to define a class template and implement its member functions outside the class body.

```cpp
#include <iostream>
using namespace std;

template <typename T>
class Calculator {
private:
    T a, b;

public:
    Calculator(T x, T y);
    T add();
};

template <typename T>
Calculator<T>::Calculator(T x, T y) {
    a = x;
    b = y;
}

template <typename T>
T Calculator<T>::add() {
    return a + b;
}

int main() {
    Calculator<int> c(10, 20);
    cout << c.add();

    return 0;
}
```

**Output**

```text
30
```

### Important syntax to remember

```cpp
template <typename T>
T Calculator<T>::add()
```

Breakdown:

- `template <typename T>` declares the template parameter.
- `Calculator<T>` identifies the specialization of the class template.
- `::add()` defines the member function belonging to that class.

The template declaration must be written before each separately defined member function template.

# 7. Non-Type Template Parameters

A template parameter does not always have to be a type. It can also be a compile-time value of an allowed type.

Example: a fixed-size array class.

```cpp
#include <iostream>
using namespace std;

template <typename T, int SIZE>
class Array {
private:
    T data[SIZE];

public:
    void setFirst(T value) {
        data[0] = value;
    }

    T getFirst() {
        return data[0];
    }
};

int main() {
    Array<int, 5> a;
    a.setFirst(100);

    cout << a.getFirst();

    return 0;
}
```

**Output**

```text
100
```

Here:

- `T` is a type parameter.
- `SIZE` is a non-type template parameter.
- `Array<int, 5>` means an array class for integers with a size of 5.

`SIZE` is known at compile time and is used to specify the array's capacity.

Modern C++ permits several kinds of non-type template parameters, subject to language rules. For an introductory OOP exam, integer parameters such as `SIZE` are a good example to learn.

# 8. Template Specialization

## 8.1 What is Specialization?

Template specialization allows you to provide a different implementation for a particular type or set of template arguments.

For example, you may want a generic `Printer` class to display most data normally, but handle `char` differently.

### Full Class Template Specialization

```cpp
#include <iostream>
using namespace std;

template <typename T>
class Printer {
public:
    void display(T value) {
        cout << "General: " << value << endl;
    }
};

template <>
class Printer<char> {
public:
    void display(char value) {
        cout << "Character: " << value << endl;
    }
};

int main() {
    Printer<int> p1;
    Printer<char> p2;

    p1.display(10);
    p2.display('A');

    return 0;
}
```

**Output**

```text
General: 10
Character: A
```

The compiler uses the general class template for `Printer<int>` and the specialized class definition for `Printer<char>`.

**Syntax to remember:**

```cpp
template <>
class ClassName<SpecificType> {
    // Specialized implementation
};
```

## 8.2 Function Template Specialization

You can also specialize a function template for a specific type.

```cpp
#include <iostream>
using namespace std;

template <typename T>
void show(T value) {
    cout << "General: " << value << endl;
}

template <>
void show<char>(char value) {
    cout << "Character: " << value << endl;
}

int main() {
    show(10);
    show('A');

    return 0;
}
```

**Output**

```text
General: 10
Character: A
```

### Specialization vs Overloading

- **Specialization:** Provides a specific implementation of a template for particular template arguments.
- **Overloading:** Provides multiple functions with the same name but different parameter lists.

These are related but different mechanisms.

# 9. Function Overloading vs Function Templates

| Function overloading | Function template |
|---|---|
| Multiple functions share the same name. | One generic function definition supports multiple types. |
| Each overload has its own implementation. | The compiler generates appropriate specializations from the template. |
| Can support type-specific logic naturally. | Best when the same logic works across types. |
| More repeated code when implementations are similar. | Reduces repeated code. |

For example, overloaded `maxValue()` functions have separate definitions for `int` and `double`, while a single function template can handle both if the comparison operation is valid.

A function template can also coexist with ordinary overloaded functions.

# 10. Advantages and Limitations of Templates

## Advantages

1. **Code reusability:** Reuse functions and classes across multiple data types.
2. **Reduced duplication:** Avoid writing the same logic repeatedly.
3. **Compile-time type checking:** The compiler checks whether the template's operations are valid for the supplied types.
4. **Generic data structures:** Useful for stacks, queues, linked lists, and trees.
5. **STL support:** Templates are fundamental to the C++ Standard Template Library.

## Limitations

1. Complex compiler errors can occur when template requirements are not satisfied.
2. Different type specializations can increase compiled code size.
3. Template syntax can be harder for beginners to understand.
4. A template works only when the supplied type supports the operations used by its implementation.
5. Template definitions are commonly placed in header files so their implementations are visible when required specializations are instantiated.

That last point is particularly useful when working on larger C++ programs.

# 11. Connection Between Templates and STL

The Standard Template Library (STL) uses templates extensively.

| STL component | Example | Purpose |
|---|---|---|
| `vector` | `vector<int> v;` | Dynamic array |
| `stack` | `stack<int> s;` | LIFO structure |
| `queue` | `queue<string> q;` | FIFO structure |
| `map` | `map<string, int> m;` | Key-value pairs |
| `sort` | `sort(v.begin(), v.end());` | Generic sorting algorithm |

Example:

```cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> v = {30, 10, 20};

    for (int x : v)
        cout << x << " ";

    return 0;
}
```

**Output**

```text
30 10 20
```

`vector<int>` is an example of a class template specialization from the STL. The `vector` class is generic, while `int` specifies the element type.

# 12. Common Mistakes in Exams and Practicals

### Mistake 1: Forgetting the template declaration

Incorrect:

```cpp
T add(T a, T b) {
    return a + b;
}
```

Correct:

```cpp
template <typename T>
T add(T a, T b) {
    return a + b;
}
```

### Mistake 2: Omitting the type in a class template declaration

```cpp
Box<int> b(10);
```

This is the usual form taught in introductory courses. Writing only `Box b(10);` relies on class template argument deduction, a modern C++ feature whose availability depends on the constructor and language version.

### Mistake 3: Using incompatible argument types

```cpp
maxValue(10, 3.5);
```

With the earlier `template <typename T> T maxValue(T, T)`, template argument deduction generally fails because the two arguments deduce conflicting types for `T`.

You could instead write:

```cpp
maxValue<double>(10, 3.5);
```

Or use two template parameters if your logic supports different types.

### Mistake 4: Assuming templates can operate on every type

A template is not automatically valid for every type. If the function uses `>`, the supplied type must support the required comparison. If it uses `+`, the supplied type must support addition.

# 13. Frequently Asked OOP Exam Questions

## Q1. Define a template in C++. Explain its advantages.

A template is a generic programming feature that allows functions or classes to work with multiple data types. Advantages include code reusability, reduced duplication, compile-time type checking, and maintainability.

## Q2. Explain function templates with a suitable example.

Define a template using `template <typename T>`, then write a generic function using `T`.

```cpp
template <typename T>
T square(T x) {
    return x * x;
}
```

Calling `square(5)` returns `25`, while `square(2.5)` returns `6.25`.

## Q3. Explain class templates with a suitable example.

A class template defines a generic class.

```cpp
template <typename T>
class Box {
    T value;

public:
    Box(T x) : value(x) {}
    T get() { return value; }
};
```

Create objects using `Box<int>` and `Box<double>`.

## Q4. Differentiate function templates and class templates.

Function templates generate generic functions for operations; class templates generate generic class specializations containing data members and member functions.

## Q5. Can a template have more than one parameter?

Yes.

```cpp
template <typename T, typename U>
class Pair {
    T first;
    U second;
};
```

It can store values of different types.

## Q6. What is template specialization?

It provides a specific implementation for particular template arguments. A full class specialization begins with `template <>` followed by the specialized class name and type.

## Q7. What are non-type template parameters?

They are compile-time values used as template arguments, such as an integer size:

```cpp
template <typename T, int N>
class Array {
    T data[N];
};
```

## Q8. What is the relationship between templates and STL?

STL containers such as `vector`, `stack`, and `queue` are class templates, while algorithms such as `sort` are function templates.
