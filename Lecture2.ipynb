## Constructor:

   - It is a special function that is invoked automatically at the time of object creation
   - Name of the constructor should as same as the class name.
   - It doesn't have any return value.
   - It is used for initialised the value.

```python
#include<iostream>
using namespace std;
class Student
{
    string name;
    int roll_no;
    int marks;

public:
    Student()
    { // Default Constructor
        cout << "Constructor is called" << endl;
        name = "Unknown";
        roll_no = 0;
        marks = 0;
    }

public:
    Student(string abc, int a, int m)
    { // parameterised Constructor
        name = abc;
        roll_no = a;
        marks = m;
    }
    // constructor overloading means we can have multiple constructors in a class with different parameters
    // what if we use parameter name same as class variables
    Student(string name, int roll_no, int marks)
    {
        this->name = name; // this is a pointer which points to the current object
        this->roll_no = roll_no;
        this->marks = marks;
    }
    // NOTE: here it will show error if we use same number of parameters in both constructors because it will not be able to differentiate between them
    // to solve this, we either can use different number of parameters in both constructors or we can use default parameters in one of the constructors
    void display()
    {
        cout << "Name: " << name << endl;
        cout << "Roll No: " << roll_no << endl;
        cout << "Marks: " << marks << endl;
    }
    //inline constructor: if we define the constructor inside the class, it is called inline constructor
    inline Student(string n, int r, int m) : name(n), roll_no(r), marks(m)
    {
    }
};
int main()
{
    Student s1; // Default Constructor will be called
    Student s2("Aryan", 101, 95); // Parameterised Constructor will be called
    Student s3("Aryan", 101, 95); // Parameterised Constructor will be called
    s1.display();
    s2.display();
    s3.display();
    return 0;
}
```

### Initialization list:
-  An `initialization list` (specifically a member initializer list) is a syntax used in constructors to initialize class data members directly before the constructor's body executes.

#### Benefits?
##### 1. Garbage Value:
- Normal Tareeke mein variable ke saath garbage value bhi initialised hoti h and  then constructor fir value assign karta h : => do baar kaam ho raha.
-  But in `Initialisation list` mein variable initialise hi value ke saath ho jaate hain.
##### 2. Performance Boost:
- Normal data types (jaise int, float) mein toh zyada farq nahi padta, lekin agar aapki class ke andar koi doosri badi Class ka object ho, toh normal assignment se pehle uska default constructor call hoga, fir assignment operator call hoga. Initialization list se wo sirf ek baar mein directly initialize ho jata hai, jisse program fast chalta hai.
##### 3. const aur reference variables
- Agar aapki class mein koi const variable hai ya koi reference variable (&) hai, toh aap use constructor ke andar assign nahi kar sakte. Unhe isi list ke through initialize karna hi padega, warna compiler error dega.
#### k<sup>3</sup>B:
- Initialization list mein variables usi order mein initialize hote hain jis order mein wo class mein declare kiye gaye hain, na ki us order mein jis order mein aapne unhe list mein likha hai.

```python
// EXAMPLE FOR 2 BENEFIT
#include <iostream>
using namespace std;

class Address {
public:
    // Default Constructor
    Address() {
        cout << "Address: Default Constructor Called (Garbage/Empty Setup)\n";
    }

    // Parameterized Constructor
    Address(string city) {
        cout << "Address: Parameterized Constructor Called for " << city << "\n";
    }

    // Assignment Operator
    Address& operator=(const Address& other) {
        cout << "Address: Assignment Operator (=) Called\n";
        return *this;
    }
};

class Student {
    Address addr; // Student ke andar Address ka object
public:
    // Normal Assignment in Body
    Student(string c) {
        addr = Address(c); // <-- Pehle temp object bana, fir assign hua
    }
};

int main() {
    cout << "--- Creating Student using Normal Assignment ---\n";
    Student s("Delhi");
    return 0;
}
/* in this, Yahan kya nuksani hui?

Jaise hi Student ka constructor shuru hua, usne addr ke liye pehle Default Constructor call kiya (unnecessary empty memory setup).

Fir Address(c) ne ek temporary object banaya (Parameterized Constructor call hua).

Fir use addr mein daalne ke liye Assignment Operator call hua.

Total Mehnat: 2 Constructors + 1 Assignment Operator call hue!*/


// professional tareeka using initialisation list:
#include <iostream>
using namespace std;

class Address {
public:
    Address() {
        cout << "Address: Default Constructor Called\n";
    }

    Address(string city) {
        cout << "Address: Parameterized Constructor Called for " << city << "\n";
    }

    Address& operator=(const Address& other) {
        cout << "Address: Assignment Operator (=) Called\n";
        return *this;
    }
};

class Student {
    Address addr;
public:
    // Initialization List use kiya
    Student(string c) : addr(c) {
        // Body ekdam khaali hai
    }
};

int main() {
    cout << "--- Creating Student using Initialization List ---\n";
    Student s("Delhi");
    return 0;
}
/*Yahan kya fayda hua?

Koi faltu default constructor call nahi hua.

Koi temporary object nahi bana, aur koi assignment operator call nahi hua.

addr object paida hote hi seedhe "Delhi" ke sath ek hi baar mein directly initialize ho gaya!

Total Mehnat: Sirf 1 Parameterized Constructor call hua.*/
```

### Example: Rectangle Class With Methods and Properties
- Define a class Rectangle with private members int length and int width. Implement a constructor to set the dimensions. Implement two public methods: calculate_area() which returns the product of length and width, and calculate_perimeter() which returns 2 * (length + width).

```python
#include <iostream>
using namespace std;
class Rectangle {
private:
    int length;
    int width;

public:
    // Constructor
    Rectangle(int l, int w) : length(l), width(w) {}

    // Method to calculate and return the area
    int calculate_area() const {
        return length * width;
    }

    // Method to calculate and return the perimeter
    int calculate_perimeter() const {
        return 2 * (length + width);
    }
};

int main() {
    Rectangle rect(10, 5); // Create a rectangle object
    
    int area = rect.calculate_area();
    int perimeter = rect.calculate_perimeter();
    
    std::cout << "Rectangle Dimensions: " << rect.getLength() << "x" << rect.getWidth() << std::endl;
    std::cout << "Area: " << area << std::endl;
    std::cout << "Perimeter: " << perimeter << std::endl;

    return 0;
}
```

### Example: Default/Parameterized Constructor
- Implement the Rectangle class (from Exercise 1). This time, include two public constructors: a default constructor that sets both length and width to 1, and a parameterized constructor that allows setting custom values for length and width.

```python
#include <iostream>
using namespace std;
class Rectangle
{
private:
    int length;
    int width;

public:
    // 1. Default Constructor (no arguments)
    Rectangle() : length(1), width(1)
    {
        cout << "Default Rectangle created (1x1)." << endl;
    }

    // 2. Parameterized Constructor
    Rectangle(int l, int w) : length(l), width(w)
    {
        cout << "Custom Rectangle created (" << length << "x" << width << ")." << endl;
    }

    int get_area() const
    {
        return length * width;
    }
};

int main()
{
    // Uses the Parameterized Constructor
    Rectangle r1(12, 4);
    cout << "R1 Area: " << r1.get_area() << endl;

    // Uses the Default Constructor
    Rectangle r2;
    cout << "R2 Area: " << r2.get_area() << endl;

    return 0;
}
```

### Copy Constructor:
    - A copy constructor in C++ is a special constructor used to create a new object as a literal copy of an existing object of the same class. It initializes the data members of the new object using the values of the existing object's members
    - By default copy constructor is present even if you don't declare exclusively.
    - If you make your own copy constructor then compiler won't create on its own and will use yours.

```python
class Student
{
    string name;
    int roll_no;
    int marks;

public:
    Student()
    { // Default Constructor
        cout << "Constructor is called" << endl;
        name = "Unknown";
        roll_no = 0;
        marks = 0;
    }
    Student(Student &s)
    { // Copy Constructor
        name = s.name;
        roll_no = s.roll_no;
        marks = s.marks;
    }
    void display()
    {
        cout << "Name: " << name << endl;
        cout << "Roll No: " << roll_no << endl;
        cout << "Marks: " << marks << endl;
    }
    
};
int main()
{
    Student s1; // Default Constructor will be called
    Student s2("Aryan", 101, 95); // Parameterised Constructor will be called
    Student s3 = s2; // Copy Constructor will be called
    Student s4(s2); // Copy Constructor will be called
    s1.display();
    s2.display();
    s3.display();
    s4.display();
    return 0;
}
```

#### Example:  Copy Constructor
- Create a class Point with a private member int x and int y as a coordinates. Implement a copy constructor for this class. Write a display() method. Demonstrate its use by creating a new Point object as a copy of an existing one.

```python
#include <iostream>
using namespace std;
class Point{
    private:
        int x;
        int y;
    public:
    Point(int a, int b):x(a), y(b){}
    Point(Point &p){
        x = p.x;
        y = p.y;
    }
    void display(){
        cout << "X : " << x << endl;
        cout << "Y : " << y << endl;
    }
    void move(int dx, int dy){
        x+=dx;
        y+=dy;
    }
};

int main()
{
    Point p1(19, 20);
    cout << "P1: " << endl;
    p1.display();
    cout << endl;
    Point p2(p1);
    cout << "P2" << endl;
    p2.display();
    p2.move(5,5);
    cout<<"After moving"<<endl;
    p1.display();
    p2.display();// p2 is deep copy of p1
}
```

# Destructor:
  - It is an instance member function which is invoked automatically when the object goes out of scope or is explicitly deleted.
  - It is used to release resources that the object may have acquired during its lifetime, such as memory or file handles. 
  - The destructor has the same name as the class but is preceded by a tilde (~) symbol. It does not take any parameters and does not return any value.

```python
#include <iostream>
using namespace std;
class Customer{
    public:
    string name;
    int *data;
    Customer(){
        name = "unkown";
        data = new int(200);
        *data = 300;
        cout << "constructor is called" << endl;
    }
    ~Customer(){
        delete data;
        cout << "End of object" << endl;
    }
};
int main(){
    Customer a;
}
```

Use a local object inside a function to clearly show when the destructor is called.

```python
#include <iostream>
using namespace std;
class Resource{
    private:
        string id;
    public:
    Resource(string i):id(i){
        cout << "Resource " << id << " is aquired" << endl;
    }
    ~Resource(){
        cout << "Rsource " << id << " is released" << endl;
    }
};
void scope_test(){
    cout << "\n-- Entering scope test--" << endl;
    Resource local("Local A"); // object created;
    cout << "\n-- Exiting scope test-- " << endl; // object(local) destroyed here;
}
int main()
{
    cout << "-- Main test -- " << endl;
    Resource main("Main _ A"); // object created;
    scope_test();
    cout << "-- Exiting main -- " << endl;
    // object (main ) destroyed here;
}
```

# Order of constructor and Destructor being called:
 - #### Why Destructor works in reverse order?
      - ##### Specific Order of Destruction
          - According to the ISO C++ FAQ and Microsoft Learn, the standard sequence for a single object is:
          - `Destructor Body:` The code inside the derived class destructor runs first.
          - `Member Objects:` Data members are destroyed in the reverse order of their declaration in the class.
          - `Base Classes:` Base class destructors run last, in the reverse order of their inheritance.
     - The primary reason destructors execute in the reverse order of construction is to protect dependencies and ensure resource safety.
     - In C++, objects are constructed from the "ground up"—starting with the base class, then moving to members, and finally the derived class. Destruction follows a "last-in, first-out" (LIFO) logic for several critical reasons:
     - <b>Dependency Protection:</b> A derived class often relies on resources (like memory, file handles, or open connections) provided by its base class. If the base class were destroyed first, the derived class destructor might attempt to access a resource that no longer exists, leading to "garbage" data or crashes.
     - <b>Logical Hierarchy:</b> Think of it like building a house: you must build the foundation (base) before the walls (derived). When tearing it down, you must remove the walls first so the foundation remains stable until it is the only thing left.
     - <b>RAII Principles:</b> This reverse order is a cornerstone of Resource Acquisition Is Initialization (RAII). For example, if you open a file in object A and start writing to it with object B, object B must finish writing (be destroyed) before object A closes the file
     - <b>Stack-Like Management:</b> Local variables on the stack are naturally destroyed in reverse order because the last object pushed onto the stack is the first one "popped" or removed when the scope ends.

```python
#include <iostream>
using namespace std;
class Customer
{
public:
    string name;
    Customer(){
        name = "D";
        cout << "Constructor is " << name << endl;
    }
    Customer(string name)
    {
        this->name = name;
        cout << "Constructor is " << name << endl;
    }
    ~Customer()
    {
        cout << "Destructor is " << name << endl;
    }
};
int main()
{
    Customer a("A"), b("B"), c("C");
    //for dynamically allocated
    Customer *d = new Customer;
    delete d;
}
// Output:
// Constructor is A
// Constructor is B
// Constructor is C
// Constructor is D
// Destructor is D
// Destructor is C
// Destructor is B
// Destructor is A
```
