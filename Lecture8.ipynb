# C++ Streams and File Handling — Detailed OOP Notes

## 1. What Is a Stream?

A **stream** is a flow of data between a program and an input/output source or destination.

- An **input stream** carries data into the program.
- An **output stream** carries data out of the program.

For example, when you type a number using the keyboard, data flows into the program. When the program displays a result on the screen, data flows out.

### Stream concept

```text
Keyboard  ---> Input stream ---> Program
Program   ---> Output stream ---> Screen
```

C++ uses streams to provide a consistent way to perform input and output.

## 2. Standard Streams in C++

The `<iostream>` header provides the common standard stream objects.

| Stream | Purpose | Example |
|---|---|---|
| `cin` | Standard input, usually the keyboard | `cin >> x;` |
| `cout` | Standard output, usually the screen | `cout << x;` |
| `cerr` | Standard error output, usually unbuffered | `cerr << "Error";` |
| `clog` | Standard error/log output, usually buffered | `clog << "Log";` |

Example:

```cpp
#include <iostream>
using namespace std;

int main() {
    int age;

    cout << "Enter age: ";
    cin >> age;

    cout << "Age = " << age << endl;

    return 0;
}
```

If the user enters `18`, output is:

```text
Enter age: 18
Age = 18
```

### `cout`, `cerr`, and `clog`

- `cout` is used for ordinary output.
- `cerr` is commonly used for errors and is not buffered in the usual implementation.
- `clog` is commonly used for diagnostic or log messages and is buffered.

All three are output streams, but they are intended for different kinds of messages.

## 3. Insertion and Extraction Operators

### Insertion operator `<<`

The insertion operator sends data to an output stream.

```cpp
cout << "Hello";
cout << 25;
```

It can be chained:

```cpp
cout << "Age: " << 18 << endl;
```

### Extraction operator `>>`

The extraction operator reads formatted data from an input stream.

```cpp
int number;
cin >> number;
```

It can also be chained:

```cpp
int a, b;
cin >> a >> b;
```

**Important:** For string input, `cin >> name` stops at whitespace. Use `getline()` to read a full line containing spaces.

## 4. Formatted and Unformatted I/O

### Formatted I/O

Formatted I/O interprets data according to its type or formatting settings.

Examples:

```cpp
int x;
cin >> x;
cout << x;
```

Common manipulators include:

- `endl` — inserts a newline and flushes the stream.
- `setw(n)` — sets the width for the next formatted field.
- `setprecision(n)` — controls floating-point precision.
- `fixed` — uses fixed-point floating-point notation.

These formatting manipulators are generally declared in `<iomanip>` where applicable.

Example:

```cpp
#include <iostream>
#include <iomanip>
using namespace std;

int main() {
    double value = 12.34567;

    cout << fixed << setprecision(2) << value << endl;
    cout << setw(8) << 42 << endl;

    return 0;
}
```

Output:

```text
12.35
      42
```

### Unformatted I/O

Unformatted I/O transfers characters or blocks of bytes without the usual formatted conversion.

Examples include:

- `get()` — reads a character.
- `put()` — writes a character.
- `getline()` — reads characters into a buffer up to a delimiter (there is also a `std::getline` free function for strings).
- `read()` — reads a specified number of bytes.
- `write()` — writes a specified number of bytes.

Example:

```cpp
#include <iostream>
using namespace std;

int main() {
    char ch;

    cout << "Enter a character: ";
    cin.get(ch);

    cout.put(ch);
    cout.put('\n');

    return 0;
}
```

## 5. Stream State and Error Checking

Streams maintain state flags that indicate whether input/output operations are succeeding.

| Function | Meaning |
|---|---|
| `good()` | No stream error flags are set. |
| `eof()` | End-of-file has been encountered. |
| `fail()` | An operation failed, often because formatted input was invalid. |
| `bad()` | A serious stream error occurred. |
| `clear()` | Clears stream error flags. |

Example:

```cpp
int value;
if (cin >> value) {
    cout << "Valid integer: " << value;
} else {
    cout << "Input failed";
}
```

A useful rule is to test the operation itself rather than checking `eof()` before reading.

```cpp
while (cin >> value) {
    cout << value << endl;
}
```

If an input operation fails, the stream state must generally be cleared before further input can succeed. If invalid characters remain in the input buffer, they may also need to be removed.

## 6. What Is File Handling?

**File handling** in C++ is the process of creating, opening, reading, writing, appending, and closing files using a program.

Variables store data temporarily in memory. Files allow data to be stored on secondary storage so it can be used later.

### Why do we need file handling?

- To store data permanently.
- To read previously saved information.
- To update or append data to an existing file.
- To handle large amounts of data without keeping everything in memory.
- To exchange data between programs.

Examples include student records, employee details, marksheets, logs, and configuration files.

## 7. File Stream Classes

C++ provides file stream classes in the `<fstream>` header.

```cpp
#include <fstream>
```

| Class | Purpose | Common use |
|---|---|---|
| `ifstream` | Input file stream | Reading from a file |
| `ofstream` | Output file stream | Writing to a file |
| `fstream` | File stream for input and output | Reading and writing |

Example declarations:

```cpp
ifstream inFile;
ofstream outFile;
fstream file;
```

## 8. Basic Steps in File Handling

1. Include `<fstream>`.
2. Declare a file stream object.
3. Open the file.
4. Check whether the operation succeeded.
5. Read from or write to the file.
6. Close the file when finished.

Example:

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    ofstream file("example.txt");

    if (!file.is_open()) {
        cerr << "Could not open file";
        return 1;
    }

    file << "Hello, file!" << endl;
    file.close();

    return 0;
}
```

Opening with the default `ofstream` output mode normally truncates an existing file, so its old contents may be erased.

## 9. Opening a File

A file can be opened using the stream constructor or the `open()` member function.

### Method 1: Constructor

```cpp
ifstream file("data.txt");
```

### Method 2: `open()`

```cpp
ifstream file;
file.open("data.txt");
```

Check whether opening succeeded:

```cpp
if (!file.is_open()) {
    cerr << "File could not be opened";
}
```

`is_open()` reports whether the stream has an associated open file.

## 10. File Opening Modes

Opening modes control how a file is opened. They are specified using `ios` flags.

| Mode | Meaning |
|---|---|
| `ios::in` | Open for input (reading). |
| `ios::out` | Open for output (writing). |
| `ios::app` | Every write goes to the end of the file. |
| `ios::ate` | Initially position at the end, but seeking elsewhere is possible. |
| `ios::trunc` | Discard existing contents when opening, where applicable. |
| `ios::binary` | Open in binary mode rather than text mode. |

Combine modes using bitwise OR (`|`):

```cpp
fstream file("data.txt", ios::in | ios::out);
```

This requests input and output access. With this particular combination, the file generally needs to exist for opening to succeed.

### Difference between `ios::app` and `ios::ate`

- `ios::app`: Every write is forced to the end of the file.
- `ios::ate`: The initial position is at the end, but you can seek to another position.

This is a frequently tested distinction.

## 11. Writing to a File Using `ofstream`

The insertion operator `<<` writes formatted data to an output file stream.

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    ofstream file("student.txt");

    if (!file) {
        cerr << "Error opening file";
        return 1;
    }

    file << "Rahul" << endl;
    file << 85 << endl;
    file << 91.5 << endl;

    file.close();

    cout << "Data written successfully";
    return 0;
}
```

Console output:

```text
Data written successfully
```

Contents of `student.txt`:

```text
Rahul
85
91.5
```

To preserve existing content and add new records, use append mode.

## 12. Reading from a File Using `ifstream`

The extraction operator `>>` reads formatted values from a file. It typically stops reading a string at whitespace.

Suppose `data.txt` contains:

```text
100 200 300
```

Program:

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    ifstream file("data.txt");

    if (!file) {
        cerr << "Error opening file";
        return 1;
    }

    int a, b, c;
    file >> a >> b >> c;

    cout << a << " " << b << " " << c;

    file.close();
    return 0;
}
```

Output:

```text
100 200 300
```

### Reading until the file ends

```cpp
int value;

while (file >> value) {
    cout << value << endl;
}
```

The loop stops when extraction fails, such as at end-of-file or when the next input is not a valid integer.

Avoid:

```cpp
while (!file.eof()) {
    file >> value;
    cout << value;
}
```

Checking `eof()` before attempting a read can process stale data one extra time. Test the read operation instead.

## 13. Reading a Whole Line Using `getline()`

The `std::getline()` function reads a complete line, including spaces, until the newline delimiter.

```cpp
#include <iostream>
#include <fstream>
#include <string>
using namespace std;

int main() {
    ifstream file("message.txt");

    if (!file) {
        cerr << "Error opening file";
        return 1;
    }

    string line;
    while (getline(file, line)) {
        cout << line << endl;
    }

    file.close();
    return 0;
}
```

If `message.txt` contains:

```text
Welcome to C++ file handling.
This is the second line.
```

Output:

```text
Welcome to C++ file handling.
This is the second line.
```

### `>>` versus `getline()`

| `file >> variable` | `getline(file, string)` |
|---|---|
| Formatted extraction. | Reads a whole line. |
| Reading a string stops at whitespace. | Spaces within the line are retained. |
| Useful for numbers and individual tokens. | Useful for sentences and full names. |

### Common `getline()` issue

If `cin >> number` is followed by `getline(cin, line)`, the newline left in the input buffer may cause `getline()` to return an empty line.

Use:

```cpp
#include <limits>

cin >> number;
cin.ignore(numeric_limits<streamsize>::max(), '\\n');
getline(cin, line);
```

This discards the remainder of the current input line before reading a full line.

## 14. Appending Data to a File

Appending adds new content at the end without discarding existing contents.

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    ofstream file("log.txt", ios::app);

    if (!file) {
        cerr << "Error opening file";
        return 1;
    }

    file << "New log entry" << endl;
    file.close();

    return 0;
}
```

`ios::app` is useful for logs, records, and histories that should retain earlier data.

## 15. Using `fstream` for Both Reading and Writing

`fstream` supports both input and output, depending on the modes used.

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    fstream file("record.txt", ios::in | ios::out);

    if (!file) {
        cerr << "Could not open file";
        return 1;
    }

    // Perform input/output operations here.

    file.close();
    return 0;
}
```

For this example, `record.txt` should already exist. If you need to create a new file, create it with an output stream first, close it, and then open it for both input and output.

When switching between reading and writing on the same update stream, follow the stream's positioning and synchronization rules. For beginner exercises, separate read and write streams are often easier to manage.

## 16. Closing a File

Use the `close()` member function:

```cpp
file.close();
```

Closing a file:

- Releases the associated file resource.
- Helps ensure buffered output is flushed.
- Allows other operations to access the file as needed.

File stream objects also close their associated files automatically when their destructors run. Explicitly calling `close()` is still useful for clarity.

## 17. File Position Pointers and Random Access

**Random access** means moving to a specific position in a file instead of reading everything sequentially.

Two position concepts are commonly discussed:

- **Get pointer:** Position used for reading.
- **Put pointer:** Position used for writing.

| Function | Purpose |
|---|---|
| `seekg()` | Moves the get (read) position. |
| `seekp()` | Moves the put (write) position. |
| `tellg()` | Reports the current get position. |
| `tellp()` | Reports the current put position. |

Example positioning:

```cpp
file.seekg(0, ios::beg);  // Read position: beginning
file.seekp(0, ios::end);  // Write position: end
```

Reference points:

- `ios::beg` — beginning of file.
- `ios::cur` — current position.
- `ios::end` — end of file.

### Example: File size using `seekg()` and `tellg()`

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    ifstream file("data.txt", ios::binary);

    if (!file) {
        cerr << "Error opening file";
        return 1;
    }

    file.seekg(0, ios::end);
    streampos size = file.tellg();

    cout << "File size: " << size << " bytes";

    file.close();
    return 0;
}
```

This illustrates a common file-size technique. Production code should also check whether seeking and `tellg()` succeeded.

## 18. Text Files vs Binary Files

| Text file | Binary file |
|---|---|
| Stores data in a human-readable text representation. | Stores bytes without requiring a text representation. |
| Can usually be inspected in a text editor. | Usually needs a suitable program to interpret it. |
| Often uses `<<`, `>>`, and `getline()`. | Often uses `read()` and `write()`. |
| Numeric values are usually stored as text characters. | Data can be stored in binary representations. |
| Text mode may translate line endings on some platforms. | Binary mode avoids text-mode translations. |

Open a file in binary mode:

```cpp
ofstream file("data.bin", ios::binary);
```

Binary mode does not automatically make a file portable between machines. Raw memory representations can differ due to type sizes, byte order, padding, and compiler details.

## 19. `read()` and `write()` for Binary Files

The unformatted member functions `read()` and `write()` transfer a specified number of bytes.

```cpp
file.write(reinterpret_cast<const char*>(&obj), sizeof(obj));
file.read(reinterpret_cast<char*>(&obj), sizeof(obj));
```

These operations are suitable for some trivially copyable objects, but raw object bytes are not a general-purpose serialization solution. Objects containing pointers, `std::string`, or other resource-owning members cannot safely be persisted this way and reconstructed as ordinary objects.

### Example with an integer

```cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    int number = 12345;

    ofstream out("number.bin", ios::binary);
    if (!out) {
        cerr << "Cannot open output file";
        return 1;
    }

    out.write(reinterpret_cast<const char*>(&number), sizeof(number));
    out.close();

    int result = 0;
    ifstream in("number.bin", ios::binary);
    if (!in) {
        cerr << "Cannot open input file";
        return 1;
    }

    in.read(reinterpret_cast<char*>(&result), sizeof(result));
    if (!in) {
        cerr << "Could not read number";
        return 1;
    }

    cout << result;

    in.close();
    return 0;
}
```

Output:

```text
12345
```

## 20. File Errors and Checking Stream Status

Always check whether a file opened and whether important operations succeeded.

```cpp
ifstream file("data.txt");

if (!file.is_open()) {
    cerr << "Unable to open data.txt";
    return 1;
}
```

Useful stream-state functions include:

| Function | Meaning |
|---|---|
| `good()` | No error flags are set. |
| `eof()` | End-of-file has been encountered. |
| `fail()` | An operation failed, often because formatted input was invalid. |
| `bad()` | A serious stream error occurred. |
| `clear()` | Clears stream error flags. |

Example:

```cpp
int value;

while (file >> value) {
    cout << value << endl;
}

if (file.bad()) {
    cerr << "A serious read error occurred";
}
```

Use the stream's Boolean state or the result of an I/O operation to test success. `eof()` becomes true only after an operation encounters end-of-file.

## 21. Random Access Example: Updating a Record

For a file containing fixed-size records, it can be possible to seek to a record's position. The offset is typically:

```text
offset = record_index * sizeof(record)
```

This simple calculation assumes fixed-size records and a compatible binary representation. It is not appropriate for variable-length text records.

For example, if a file stores an array of fixed-size integers:

```cpp
file.seekg(index * sizeof(int), ios::beg);
```

Use binary mode and validate the index and operation results. In real applications, portable serialization is preferable to relying on raw in-memory layouts for complex objects.

## 22. Streams vs File Handling

| Streams | File handling |
|---|---|
| General mechanism for transferring data. | Uses streams to work with files specifically. |
| Includes console streams such as `cin` and `cout`. | Uses file streams such as `ifstream`, `ofstream`, and `fstream`. |
| Defined across stream-related headers, including `<iostream>`. | File stream classes are provided by `<fstream>`. |
| May communicate with a keyboard, screen, or other source/destination. | Reads and writes data in files. |

**Key relationship:** File handling in C++ is built on the stream model.

## 24. Common Mistakes in Exams and Practicals

### Mistake 1: Forgetting `<fstream>`

File stream classes require:

```cpp
#include <fstream>
```

### Mistake 2: Using `ofstream` without considering truncation

```cpp
ofstream file("data.txt");
```

This normally overwrites existing contents. Use `ios::app` when you want to append.

### Mistake 3: Using `while (!file.eof())`

Incorrect:

```cpp
while (!file.eof()) {
    file >> value;
    cout << value;
}
```

Correct:

```cpp
while (file >> value) {
    cout << value << endl;
}
```

### Mistake 4: Using `>>` to read a full name

```cpp
cin >> name;
```

This reads only up to the first whitespace. Use `getline()` for full names or sentences.

### Mistake 5: Forgetting to check whether a file opened

Always test the stream before using it.

### Mistake 6: Confusing `ios::app` and `ios::ate`

Remember: `app` forces writes to the end; `ate` only sets the initial position to the end.

## 25. Frequently Asked OOP Exam Questions

### Q1. What is a stream in C++?

A stream is a flow of data between a program and an input or output source. Input streams bring data into a program, while output streams send data out.

### Q2. Explain `cin`, `cout`, `cerr`, and `clog`.

- `cin`: Standard input.
- `cout`: Standard output.
- `cerr`: Error output, commonly unbuffered.
- `clog`: Logging or diagnostic output, commonly buffered.

### Q3. What are the main file stream classes?

`ifstream` is used for reading, `ofstream` for writing, and `fstream` for both input and output.

### Q4. Explain file opening modes.

`ios::in` reads, `ios::out` writes, `ios::app` appends every write, `ios::ate` initially positions at the end, `ios::trunc` truncates existing content, and `ios::binary` selects binary mode.

### Q5. Differentiate text and binary files.

Text files store data as readable text, whereas binary files store bytes in a representation interpreted by a program. Binary mode avoids text-mode translations, but raw binary layouts are not automatically portable.

### Q6. Explain `seekg()`, `seekp()`, `tellg()`, and `tellp()`.

`seekg()` changes the read position, `seekp()` changes the write position, `tellg()` reports the read position, and `tellp()` reports the write position.

### Q7. What is the difference between `ios::app` and `ios::ate`?

`ios::app` forces each write to the end of the file. `ios::ate` starts at the end but permits seeking elsewhere.

### Q8. Why should `while (!file.eof())` generally be avoided?

The EOF flag is set after an input operation encounters end-of-file. Checking it before the read can cause stale data to be processed. Prefer `while (file >> value)` or `while (getline(file, line))`.

### Q9. What are `read()` and `write()`?

They are unformatted stream member functions used to transfer a specified number of bytes, commonly for binary I/O.

### Q10. How do streams relate to file handling?

Streams provide the general input/output mechanism. File handling uses file stream classes from `<fstream>` to transfer data to and from files.
